```python
dados.reset_index(inplace=True, drop=True)
dados.head()
```

```python
import numpy as np
```

```python
dados['max_hospedes'] = dados['max_hospedes'].astype(np.int64)
display(dados['max_hospedes'])
```

```python
colunas_numericas = ["quantidade_banheiros", "quantidade_quartos", "quantidade_camas"]
dados[colunas_numericas] = dados[colunas_numericas].astype(np.int64)

display(dados[colunas_numericas])
```

```python
dados.info()
```

```python
dados.head()
```

```python
dados['preco']
```

```python
dados['preco'] = dados['preco'].apply(lambda x: x.replace('$', '').replace(',', ''))
```

```python
dados['preco'] = dados['preco'].astype(np.float64)
```

```python
display(dados.head())
```

```python
dados[['taxa_deposito', 'taxa_limpeza']] = dados[['taxa_deposito', 'taxa_limpeza']].map(
    lambda x: float(x.replace('$', '').replace(',', '').strip())
)
```

```python
dados[['taxa_deposito', 'taxa_limpeza']].astype(np.float64)
```

```python
dados.head()
```

```python
dados["descricao_local"] = dados["descricao_local"].str.lower()

dados.info()
```

```python
dados["descricao_local"]
```

```python
dados["descricao_local"] = dados["descricao_local"].str.replace('[^a-zA-Z0-9\-\']', ' ', regex=True)

dados["descricao_local"] = dados["descricao_local"].str.replace('(?<!\w)-(?!\w)', '', regex=True)
```

```python
dados.head()
```

```python
dados["comodidades"] = dados["comodidades"].str.replace('\{|}|\"', "", regex=True)

dados["comodidades"] = dados["comodidades"].str.split(",")
```

```python
dados["comodidades"]
```

```python
display(dados.head())
```

```python
dados["descricao_vizinhanca"] = dados["descricao_vizinhanca"].str.lower()
```

```python
dados["descricao_vizinhanca"] = dados["descricao_vizinhanca"].str.replace('\{|}|\"', "", regex=True)

dados["descricao_vizinhanca"] = dados["descricao_vizinhanca"].str.replace('(?<!\w)-(?!\w)', '', regex=True)
```
