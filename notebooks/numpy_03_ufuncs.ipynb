```python
import numpy as np
import pandas as pd

dados = {
    "aluno": ["Ana", "Bruno", "Carla", "Diego", "Eva", "Felipe"],
    "prova1": [8.5, 6.0, 9.2, 4.5, 7.0, 5.5],
    "prova2": [7.0, 6.5, 8.8, 5.0, 7.5, 6.0],
    "prova3": [9.0, 5.5, 9.5, 6.0, 8.0, 4.0],
    "prova4": [8.0, 7.0, 9.0, 5.5, 6.5, 5.0]
}

df = pd.DataFrame(dados)
```

```python
df["media"] = df[["prova1", "prova2", "prova3", "prova4"]].mean(axis=1)
display(df)
```

```python
condicoes = [
    df["media"] >= 7,
    (df["media"] >= 5) & (df["media"] < 7),
]

resultado = [
    "aprovado",
    "recuperacao",
]
df["situacao"] = np.select(
    condicoes,
    resultado,
    default="reprovado"
)
display(df)
```

```python
aprovados = df[
    df["situacao"] == "aprovado"
]

display(aprovados)
```

```python
maior_media = aprovados["media"].max()
display(maior_media)
```
