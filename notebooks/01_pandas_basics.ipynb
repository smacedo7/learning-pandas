```python
import pandas as pd
import numpy as np

from pathlib import Path

print(Path.cwd())
```

```python
df_vendas = pd.read_csv("../datasets/vendas_tech.csv", low_memory=False)
display(df_vendas)

df_gerentes = pd.read_excel('../datasets/gerentes_lojas.xlsx')
display(df_gerentes)

display(df_vendas.info())
display(df_vendas.describe())
```

```python
df_analysis = df_vendas.drop(columns=['Data_Base'])

df_analysis["Loja"] = df_analysis["Loja"].fillna("Online")

df_analysis["Data"] = pd.to_datetime(df_analysis["Data"], format='%Y-%m-%d')

df_analysis["Loja"] = df_analysis["Loja"].str.strip()
df_analysis["Loja"] = df_analysis["Loja"].str.title()

df_analysis = df_analysis.drop_duplicates(subset=["ID_Pedido"])



display(df_analysis.info())
display(df_analysis.columns)
display(df_analysis.head())
display(df_analysis)
```

```python
df_analysis['Faturamento'] = df_analysis["Preco_Unitario"] * df_analysis["Qtd"]

df_analysis["Forma_de_Venda"] = np.where(
    df_analysis["Loja"]=="Online",
    "Online",
    "Presencial"
)

regions_dict = {
    'São Paulo': 'Sudeste',
    'Belo Horizonte': 'Sudeste',
    'Online': 'Online',
    'Rio De Janeiro': 'Sudeste',
    'Curitiba': 'Sul',
    'Porto Alegre': 'Sul',
    'Recife': 'Nordeste',
    'Salvador': 'Nordeste'
}

df_analysis["Regiao"] = df_analysis["Loja"].map(regions_dict)

display(df_analysis)
```

```python
# analisar -> filtrar dados

corrigido = df_analysis.sort_values(by='Data')  # -> esta mudando pq existem varias datas do mesmo dia

# .loc[] -> .loc[linha, 'Nome da coluna']
loja = corrigido.loc[3, 'Loja']  
cliente = corrigido.loc[4, 'Cliente']
produto = corrigido.loc[4, 'Produto']

# .iloc -> por posicao (indice, coluna, sem contar coluna de indices) -> .iloc[linha, coluna]
id_pedido = 2
loja_id_2 = corrigido.iloc[2, 2]
cliente_id_2 = corrigido.iloc[2, 6]
produto_id_2 = corrigido.iloc[2, 3]

# condicional: 
df_id_pedido_4 = corrigido[corrigido['ID_Pedido'] == 4]

# exportar as vendas de 2024
df_exportar_vendas_2024 = corrigido[corrigido['Data']=="2024-01-01"]

# exportar pedacos da base: 
df_vendas_sp = corrigido[corrigido["Loja"]=="São Paulo"]
df_vendas_sp.to_csv("Vendas_sp.csv", index=False)

# dupla comdicional
df_vendas_HDMI_SUL = corrigido[
    (corrigido['Produto'] == "Cabo HDMI") &
    (corrigido['Regiao'] == "Sul")
]

display(corrigido)

print(loja, cliente, produto)
print(loja_id_2, cliente_id_2, produto_id_2)

display(df_id_pedido_4)
display(df_exportar_vendas_2024)

display(df_vendas_HDMI_SUL)


```

```python
# analise por agrupamento: 

# ranking de faturamento por loja: 
analise_lojas = corrigido[["Loja", "Faturamento"]].groupby("Loja").sum()
analise_lojas = analise_lojas.sort_values(by="Faturamento", ascending=False)
analise_lojas = analise_lojas.reset_index()  # -> faz os indices voltarem
analise_lojas["Faturamento"] = analise_lojas["Faturamento"].map("RS: {:,.2f}".format)

# ranking de produtos que mais venderam online
vendas_online = corrigido[corrigido["Loja"]=="Online"]
online = vendas_online[["Faturamento", "Qtd"]].groupby("Faturamento").sum()
online = online.sort_values(by="Qtd", ascending=False)
online = online.rename(columns={"Qtd": "Vendas Totais"})

# quais produtos venderam mais em cada uma das lojas:
analise_mais_vendidos = corrigido[["Loja", "Produto", "Qtd"]].groupby(["Loja", "Produto"]).sum()

display(analise_lojas)
display(online)
display(analise_mais_vendidos)
```
