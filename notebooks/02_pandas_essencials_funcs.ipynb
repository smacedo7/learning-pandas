```python
import pandas as pd

dados = {
    'Nome': ['Ana', 'Bruno', 'Carlos', 'Daniela', 'Eduardo'],
    # 'Area': ['Administrativo', 'Financeiro', 'Vendas', 'Administrativo', 'Financeiro'],
    'Idade': [25, 30, 35, 40, 45],
    'Salário': [3000, None, 7000, None, 11000]
}

df = pd.DataFrame(dados)
```

```python
# describe function: 

display(df.describe())
```

```python
# apply function

df["Salario Ajustado"] = df["Salário"].apply(lambda x: x * 1.1)

display(df)
```

```python
# groupby

# medias_area = df.groupby('Area').mean(numeric_only=True)
# display(medias_area)
```

```python
# fillna -> preenche valores nulos

media_salarial = df["Salário"].mean(numeric_only=True)
df["Salário"] = df["Salário"].fillna(media_salarial)

df["Salario Ajustado"] = df["Salario Ajustado"].fillna(media_salarial)


display(df)
```

```python
# merge: mesclar tabelas

tabela_areas = pd.DataFrame({
    "Nome Funcionario": ['Ana', 'Bruno', 'Carlos', 'Eduardo'],
    "Area": ['Administrativo', 'Financeiro', 'Vendas', 'Financeiro']
})

df = df.merge(tabela_areas, how='left', left_on='Nome', right_on='Nome Funcionario')
df = df.drop(columns="Nome Funcionario")

display(df)
display(tabela_areas)


```
