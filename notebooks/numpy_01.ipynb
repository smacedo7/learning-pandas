```python
import numpy as np
import pandas as pd
```

```python
# -> array multidimensional usando lista de listas

multidimensional_array = np.array([range(i, i+3) for i in [2, 4, 6]])
display(multidimensional_array)

# array([[2, 3, 4],
#        [4, 5, 6],
#        [6, 7, 8]])
# linhas internas tratadas como linhas de um array
# bidimensional resultante
```

```python
# create an array of integers with lenght = 10 and field with 0

zero_arr = np.zeros(10, dtype=int)
display(zero_arr)
```

```python
# array with only one values, 3x5 lenght

one_arr = np.ones((3, 5), dtype=int)
display(one_arr)
```

```python
# create a full array with value = 3.14

full_array = np.full((3, 5), 3.14)
display(full_array)
```

```python
# like the range func, create an array with a linear sequence

range_arr = np.arange(0, 20, 2)
display(range_arr)
```

```python
# array equal spaced in a determined range

equal_arr = np.linspace(0, 1, 5, dtype=float)
display(equal_arr)
```

```python
# random array between 0 and 1 

r_arr = np.random.random((3, 3))  # pass through only one argument
display(r_arr)
```

```python
# create a random array with means and standard deviation

mat_arr = np.random.normal(0, 1, (3,3)) 
# mean = 0
# standart deviation = 1 (*desvio padrao)

display(mat_arr)
```

```python
# create a random array with random numbers

random_arr = np.random.randint(0, 10, (3, 3))

display(random_arr)
```

```python
# create an indentity matriz

matrix_arr = np.eye(3)
display(matrix_arr)
```

```python
# create a not innitializated array with numbers in that memory allocation

empty_arr = np.empty(3)
display(empty_arr)
```
